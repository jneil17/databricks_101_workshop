*Share your F1 analytics with business stakeholders through an intuitive, simplified interface*

## 🎯 What is Genie?

Genie is a **simplified business user interface** that provides a single entry point for consuming data and AI assets without technical complexity. Perfect for sharing the F1 analytics you've built!

### 🏆 **Your F1 Workshop Assets**
Throughout this workshop, you've created:
* 📊 **AI/BI Dashboards** - F1 performance metrics and race analytics
* 🧞 **Genie Spaces** - Natural language queries for F1 data
* 📦 **Pipelines & Workflows** - Scheduled ETL jobs for daily metrics
* 📈 **SQL Queries** - Race results and driver performance analysis

### 🎯 **Why Genie?**
* **Business-friendly interface** - No technical navigation required
* **Consumer access** - Perfect for stakeholders who need insights, not development
* **Single entry point** - All your F1 assets in one place
* **Intuitive design** - Focus on insights, not infrastructure

## 🗓 Step 5: What else Genie can do (just call these out)

Beyond the dashboards + Genie spaces flow you've already seen, point attendees at two newer pieces of the business-user experience:

### Scheduled tasks (recurring Genie summaries)
Genie can run an analytical question on a schedule and deliver the result (table, chart, or summary) into the user's inbox or a Slack channel. Examples for F1:

* *"Every Monday at 8am, send me the updated driver standings."*
* *"After every race, post a summary of who scored points to #f1-watch."*

Configure these from a Genie answer → **⋯ → Schedule**.

### Domains
Domains let you group related assets — F1 dashboards + the Genie space + saved questions — under one named workspace so business users see a curated landing page instead of a free-for-all of every asset they have access to.

Example: a **Formula 1 Analytics** domain that contains the Teams dashboard, the Drivers dashboard, the F1 Championship Genie space, and a couple of pinned saved questions. Stakeholders pick the domain and get everything in one place.

**Why bring this up in the workshop:** attendees usually have one of two questions at this point — *"can my execs get pushed updates?"* (yes, scheduled tasks) and *"can I curate what they see?"* (yes, domains). A 60-second mention each is plenty.
