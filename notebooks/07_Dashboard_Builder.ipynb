# 📊 Dashboard Builder (Genie Code)

Rather than walking you through clicking each chart by hand, this notebook gives you a **single Genie Code prompt** that builds the whole Formula 1 dashboard for you on top of your gold tables.

We've kept `08_Formula_1_Dashboard.lvdash.json` in this repo as a **pre-baked reference dashboard** — if Genie's output looks different from your neighbor's, you can always import the JSON for a known-good baseline.

## 🛠 Build it with Genie Code

1. In the left sidebar, go to **Dashboards → ➕ Create**.
2. Name it `Formula 1 Workshop` and **pick a SQL warehouse** (serverless is fine — without one, Genie can't query the tables and the build will fail).
3. Open the **Genie Code** side panel (✨ icon, top right of the dashboard).
4. Paste the entire prompt below and let it build. **Expect ~1–2 minutes** while it generates queries and lays out charts.

> ⚠️ **Non-deterministic.** Genie won't produce exactly the same dashboard twice. Your neighbour's chart titles, sort orders, or chart types may differ from yours, and both will differ from the pre-baked reference dashboard. That's fine — the point is to feel the workflow.

```
Build a Formula 1 analytics dashboard with two pages.

Use these existing gold tables directly. Do NOT create metric views,
materialized views, or any other intermediate objects. Reference the
tables exactly as named below.

Data sources (Unity Catalog):
- main.default.f1_gold_driver_championship
  columns: driver, team, races_entered, total_points, wins, podiums,
           points_finishes, avg_finish_position, points_per_race
- main.default.f1_gold_team_championship
  columns: team, drivers_count, total_entries, total_points, wins, podiums,
           avg_finish_position, points_per_entry

Page 1: "Teams"
- KPI tiles row at the top: count of teams, sum of total_points,
  sum of wins, sum of podiums — each as a single large-number tile.
- Bar chart: total_points per team, sorted descending, colored by team.
- Table: full f1_gold_team_championship sorted by total_points desc.
  Columns: team, total_points, wins, podiums, points_per_entry, avg_finish_position.

Page 2: "Drivers"
- KPI tiles row at the top: count of drivers, sum of races_entered,
  sum of wins, sum of podiums.
- Bar chart: top 10 drivers by total_points, colored by team.
- Scatter plot: x = avg_finish_position (lower is better),
  y = total_points, size = wins, color = team, label = driver.
- Table: full f1_gold_driver_championship sorted by total_points desc.
  Columns: driver, team, total_points, wins, podiums, points_per_race,
           avg_finish_position.

Global filter (applies to both pages): team multi-select.

Styling: clean default theme, no gridlines on bar charts.
After building, publish the dashboard and confirm both pages render with data.
```


## 🛟 Fallback: import the pre-baked dashboard

If Genie didn't produce what you wanted, or you want a guaranteed-working version:

1. **Dashboards → ⋯ → Import dashboard from file**.
2. Pick `notebooks/08_Formula_1_Dashboard.lvdash.json` from this repo.
3. Confirm it points at `main.default.f1_gold_driver_championship` and `main.default.f1_gold_team_championship`.

If you used a different catalog (see `00_Setup.ipynb`'s catalog override note), edit the dataset queries inside the dashboard after import.

## 🧰 Want to inspect / tweak the SQL?

Every chart on a Databricks dashboard is backed by a SQL dataset. To explore or tune one:

1. Click a chart → **View dataset SQL**.
2. Click **Open in SQL editor**.
3. The SQL editor (left sidebar → **SQL → New query**) is where you can iterate, save shared queries, or run ad-hoc analytics outside the dashboard.

This is the same workflow you'll use when Genie generates a SQL answer and you want to tweak it — copy it into the SQL editor and keep going.
